# Calculating cosine similarity of two sentences

```python
import numpy as np
import ollama
```

```python
sentence_1 = "The bank account is suspended due to suspicious activity"  # "bank" refers to a financial institution
sentence_2 = "The river bank is located both in the East and West side of the country"  # "bank" refers to a river bank

query_sentence = "What should I do re-activate my bank account?"  # "bank" refers to a financial institution and therefore must be more similar/closer to sentence_1
```

Define some helper functions to get vector embeddings for textual sentences and to calucate cosine similarity between two vectors

```python
def get_embeddings(input_text, embedding_model="nomic-embed-text"):
    embeddings = ollama.embed(model=embedding_model, input=input_text).embeddings[0]
    return np.array(embeddings)

def calc_cosine_similarity(vec_a, vec_b):
    dot_product = np.dot(vec_a, vec_b)  # a.b

    norm_a = np.linalg.norm(vec_a)  # |a|
    norm_b = np.linalg.norm(vec_b)  # |b|

    return dot_product / (norm_a * norm_b)  # (a.b)/|a||b|
```

Now calculate vector embeddings for the textual sentences. The model `nomic-embed-text` is already trained to get correct embeddings

```python
sentence_1_embeddings = get_embeddings(sentence_1)
sentence_2_embeddings = get_embeddings(sentence_2)

query_sentence_embeddings = get_embeddings(query_sentence)
```

Now, we'll verify the vector embeddings by calculating cosine similarity between them

```python
cosine_similarity_sentence_1_and_query = calc_cosine_similarity(sentence_1_embeddings, query_sentence_embeddings)
cosine_similarity_sentence_2_and_query = calc_cosine_similarity(sentence_2_embeddings, query_sentence_embeddings)
```

If the model is trained correctly, it should refer "bank" as a financial institution in the query and get closer vector to sentence_1

```python
print(f"Cosine similarity of query with sentence_1 is higher: {cosine_similarity_sentence_1_and_query}")
print(f"Cosine similarity of query with sentence_2 is lower: {cosine_similarity_sentence_2_and_query}")
```
